```python
import pandas as pd
from pymongo import MongoClient

# Connect to MongoDB
client = MongoClient("mongodb://localhost:27017/") #con
db = client["FoodData"]  # Replace with your database name
collection = db["product"]  # Replace with your collection name

# Load CSV
df = pd.read_csv("data.csv")

# Convert DataFrame to JSON (MongoDB format)
data = df.to_dict(orient="records")

# Insert into MongoDB
collection.insert_many(data)

print("Data imported successfully!")
```
